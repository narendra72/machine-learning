```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df
```

```python
df.head(10)
```

#### Univariate EDA on Categorical Data 

```python
# 1. count plot -->  frequence count of each value
sns.countplot(data=df, x='Survived')  
```

```python
sns.countplot(data=df, x='Sex')
```

```python
# 2. pieplot 
df['Sex'].value_counts().plot(kind='pie', autopct='%1.1f%%')
```

#### Univariate EDA on Numrical Data 

```python
# 1. histogram   --> Axes-level
sns.histplot(data=df,x='Age', bins=10,hue='Sex',kde=True)
```

```python
# displot --> Figure-level
sns.displot(data=df,x='Age', bins=10, kde=True)   
```

```python
# box plot   --> distribution + spread + outliers dikhata hai
sns.boxplot(data=df, x='Age')
```
