## One Hot encode using Pandas

```python
pd.get_dummies(df,columns=['fuel','owner'])
```

```python
# k-1 encoding 
# to remove first column
pd.get_dummies(df,columns=['fuel','owner'],drop_first=True)
```

We do not use this encoding method directly with Pandas because Pandas does not preserve the original column positions
the column order may change after encoding.

### One Hot Encoding using Sklearn

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,0:4],df.iloc[:,-1],test_size=0.2)
```

```python
x_test.head()
```

```python
y_test.head()
```

```python
y_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohn = OneHotEncoder(drop='first',sparse_output=False)
```

```python
new_x_train = ohn.fit_transform(x_train[['fuel','owner']])
```

```python
print(new_x_train)
```

```python
new_x_test = ohn.transform(x_test[['fuel','owner']])
```

```python
print( new_x_test )
```

### One Hot Encoding on brand column 

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold=100
```

```python
repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl,'uncommon'),dtype=int)
```
